```python
np.random.seed(42)

marks = np.random.randint(20, 101, (12, 4))

print("Marks:")
print(marks)

avg = np.mean(marks, axis=1)
print("Student averages:", avg)

print("Test mean:", np.mean(marks, axis=0))

print("Test standard deviation:", np.std(marks, axis=0))

passed = np.all(marks >= 40, axis=1)

print("Students passing all tests:", np.sum(passed))

print("Students with average >= 75:")
print(np.where(avg >= 75)[0] + 1)

print("Test with greatest variation:",
      np.argmax(np.std(marks, axis=0)) + 1)
```

```python
a = np.arange(21, 46).reshape(5, 5)

print("Original matrix:")
print(a)

b = a[1:4, 1:4]

print("Central matrix:")
print(b)

c = b.T

print("Transpose:")
print(c)

d = np.dot(b, c)

print("Matrix product:")
print(d)

print("Row-wise sums:", d.sum(axis=1))

print("Maximum:", d.max())

print("Maximum position:", np.argwhere(d == d.max()))
```

```python
quantity = np.array([
    [10, 20, 15, 25],
    [18, 22, 20, 15],
    [25, 18, 22, 20],
    [20, 25, 18, 22],
    [15, 20, 25, 18],
    [22, 18, 20, 25]
])

price = np.array([150, 240, 90, 320])

# Daily revenue
daily_revenue = quantity @ price

print("Daily revenue:", daily_revenue)

# Product-wise quantity
total_quantity = quantity.sum(axis=0)

print("Product quantity:", total_quantity)

# Product-wise revenue
product_revenue = total_quantity * price

print("Product revenue:", product_revenue)

print("Total revenue:", daily_revenue.sum())

print("Highest revenue day:",
      np.argmax(daily_revenue) + 1)

print("Best selling product:",
      np.argmax(total_quantity) + 1)

# Reduce price by 10% if quantity > 100
new_price = np.where(
    total_quantity > 100,
    price * 0.90,
    price
)

new_revenue = total_quantity * new_price

print("New prices:", new_price)

print("New total revenue:", new_revenue.sum())

print("Original prices:", price)
```

```python
np.random.seed(42)

data = np.random.normal(50, 12, 100)

print("Outside values:",
      np.sum((data < 30) | (data > 70)))

cleaned = np.where(data < 30, 30,
          np.where(data > 70, 70, data))

print("Original mean:", np.mean(data))
print("Cleaned mean:", np.mean(cleaned))

print("Original variance:", np.var(data))
print("Cleaned variance:", np.var(cleaned))

print("Original std:", np.std(data))
print("Cleaned std:", np.std(cleaned))

print("Original min:", np.min(data))
print("Cleaned min:", np.min(cleaned))

print("Original max:", np.max(data))
print("Cleaned max:", np.max(cleaned))
```

```python
branch_a = pd.Series({
    'Q1': 120,
    'Q2': 145,
    'Q4': 180
})

branch_b = pd.Series({
    'Q2': 110,
    'Q3': 135,
    'Q4': 155
})

branch_c = pd.Series({
    'Q1': 95,
    'Q3': 125,
    'Q4': 140
})

# Ordinary addition
result = branch_a + branch_b + branch_c

print("Ordinary addition:")
print(result)

# Common quarters
common = branch_a.index.intersection(
    branch_b.index
).intersection(branch_c.index)

print("Common quarters:", common)

print("Common quarter earnings:")
print(branch_a[common] +
      branch_b[common] +
      branch_c[common])

# Treat missing values as zero
result2 = (
    branch_a.add(branch_b, fill_value=0)
    .add(branch_c, fill_value=0)
)

print("Missing treated as zero:")
print(result2)
```

```python
scores = {
    'R101': 78,
    'R102': 65,
    'R104': 92,
    'R105': 48
}

required = [
    'R105', 'R103', 'R101',
    'R106', 'R104', 'R102'
]

s = pd.Series(scores, index=required)

print("Original Series:")
print(s)

# Missing records
print("Missing records:")
print(s[s.isnull()])

# Valid scores >= 70
valid = s[s.notnull() & (s >= 70)]

print("Valid scores:")
print(valid)

# Convert to dictionary
d = valid.to_dict()

print("Dictionary:")
print(d)

# Reconstruct Series
new_s = pd.Series(d)

print("New Series:")
print(new_s)

# Verify
print("Verification:")
print(valid.equals(new_s))
```
